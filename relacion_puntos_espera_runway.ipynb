```python
import folium

# Crear un mapa centrado en una ubicación específica
mapa = folium.Map(location=[40.49184637554621, -3.572616577148437], zoom_start=15)  # Madrid, España

# Lista de coordenadas (latitud, longitud)
# Añadir marcadores al mapa
for icao, lat, lon, runw, fe in b:
    print(lat, lon, runw, fe)
    folium.Marker([lat, lon],tooltip=f"ICAO: {icao}, Lat: {lat}, Lon: {lon}, Fecha: {fe}", icon=folium.Icon(color=colores[runw], icon="info-sign")).add_to(mapa)

# Guardar el mapa en un archivo HTML y mostrarlo
mapa.save("mapa.html")
```

```python
sec = set()
df_sec = df.sort_values(by=["ICAO", "despegue"])
actual = None
despegue_actual = None
lista_prov=[]
for _, fila in df_sec.iterrows():
    if ((actual == None) | (fila["ICAO"]!=actual)):
        actual = fila["ICAO"]
        despegue_actual = fila["despegue"]
        if len(lista_prov)>0:
            sec.add(tuple(lista_prov))
            lista_prov = []
    if fila["despegue"] == despegue_actual:
        holding_point=fila["holding_point"]
        if len(lista_prov)>0:
            if lista_prov[-1] != holding_point:
                lista_prov.append(holding_point)
        else:
            lista_prov.append(holding_point)
    if fila["despegue"] != despegue_actual:
        sec.add(tuple(lista_prov))
        lista_prov = []
print(sec)
```
