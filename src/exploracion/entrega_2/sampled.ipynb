```python
import pandas as pd
import matplotlib.pyplot as plt
import warnings
from plotly_calplot import calplot

df['timestamp'] = pd.to_datetime(df['timestamp'])
df['date'] = df['timestamp'].dt.floor('d')
daily_avg_queue = df.groupby('date')['queue_length'].mean().reset_index()
vmin = daily_avg_queue['queue_length'].min()
vmax = daily_avg_queue['queue_length'].max()

with warnings.catch_warnings(): # pa evitar warning
        warnings.simplefilter("ignore")
        fig_2024 = calplot(daily_avg_queue[daily_avg_queue['date'].dt.year == 2024], 
                x='date', 
                y='queue_length',
                name='queue_length',
                title = '2024',
                dark_theme=True,
                showscale=True,
                colorscale="Viridis",
                start_month=11,
                month_lines_width=3,
                total_height=300,
                cmap_min=vmin, 
                cmap_max=vmax,
        )

        fig_2025 = calplot(daily_avg_queue[daily_avg_queue['date'].dt.year == 2025], 
                x='date', 
                y='queue_length',
                name='queue_length',
                title = '2025',
                dark_theme=True,
                showscale=True,
                colorscale="Viridis",
                end_month=2,
                month_lines_width=3,
                total_height=300,
                cmap_min=vmin, 
                cmap_max=vmax,
        )
print('QUEUE_LENGTH MEDIO POR DIA')
fig_2024.show()
fig_2025.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import warnings
from plotly_calplot import calplot

df['timestamp'] = pd.to_datetime(df['timestamp'])
df['date'] = df['timestamp'].dt.floor('d')
daily_avg_queue = df.groupby('date')['queue_length'].max().reset_index()
vmin = daily_avg_queue['queue_length'].min()
vmax = daily_avg_queue['queue_length'].max()

with warnings.catch_warnings(): # pa evitar warning
        warnings.simplefilter("ignore")
        fig_2024 = calplot(daily_avg_queue[daily_avg_queue['date'].dt.year == 2024], 
                x='date', 
                y='queue_length',
                name='queue_length',
                title = '2024',
                dark_theme=True,
                showscale=True,
                colorscale="Viridis",
                start_month=11,
                month_lines_width=3,
                total_height=300,
                cmap_min=vmin, 
                cmap_max=vmax,
        )

        fig_2025 = calplot(daily_avg_queue[daily_avg_queue['date'].dt.year == 2025], 
                x='date', 
                y='queue_length',
                name='queue_length',
                title = '2025',
                dark_theme=True,
                showscale=True,
                colorscale="Viridis",
                end_month=2,
                month_lines_width=3,
                total_height=300,
                cmap_min=vmin, 
                cmap_max=vmax,
        )
print('QUEUE_LENGTH MAXIMO POR DIA')
fig_2024.show()
fig_2025.show()
```

```python
counts = df.groupby('parado').queue_length.value_counts().reset_index()
counts.columns = ['parado', 'queue_length', 'count']
fig = px.bar(counts, x='queue_length', y='count', color='parado', text='count',width=500, height=500)
fig.show()
```

Vemos que el queue_length está relacionado con parado, pues ya a partir de queue_length = 1 se nota que la mayor proporción es de los parados.
